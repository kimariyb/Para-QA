## Rename the pdf names

```python
import os

# 读取 data 文件夹下的所有 pdf 文件
files = os.listdir('./data')
# 将文件按照序号重命名，0001.pdf, 0002.pdf, ...
for i, file in enumerate(files):
    os.rename(os.path.join('./data', file), os.path.join('./data', f'{i+1:04d}.pdf'))
```

## Transfer the markdown file to output folder

```python
import shutil

from tqdm import tqdm
from pathlib import Path
src_dir= Path('mineru').resolve()
dst_dir= Path('./mineru').resolve()

# 确保 dst_dir 文件夹存在
dst_dir.mkdir(parents=True, exist_ok=True)

# 读取 mineru 文件下的所有文件夹，过滤非文件夹项
mineru_folders = [f for f in src_dir.iterdir() if f.is_dir()]

# 第一阶段：收集所有 .md 文件路径
md_files = []
for folder in tqdm(mineru_folders, desc='Scanning folders'):
    # 进入 vlm 文件夹下
    vlm_dir = folder / "vlm"
    # 读取 vlm 文件夹下的所有文件
    if vlm_dir.exists():  # 检查 vlm 文件夹是否存在
        found_files = list(vlm_dir.glob('*.md'))  # 先获取匹配结果
        if found_files:
            md_files.extend(found_files)
        else:
            print(f"\nNo .md files found in: {vlm_dir}")  # 空文件夹提示

copied_count = 0
for src_file in tqdm(md_files, desc='Copying files'):
    try:
        dst_file = dst_dir / src_file.name
        if not dst_file.exists():  # 避免重复复制
            shutil.copy2(src_file, dst_file)  # copy2 保留元数据
            copied_count += 1
    except Exception as e:
        print(f"\nError copying {src_file.name}: {str(e)}")
```
