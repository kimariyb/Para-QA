## Rename the pdf names

```python
import os

# 读取 data 文件夹下的所有 pdf 文件
files = os.listdir('./data')
# 将文件按照序号重命名，0001.pdf, 0002.pdf, ...
for i, file in enumerate(files):
    os.rename(os.path.join('./data', file), os.path.join('./data', f'{i+1:04d}.pdf'))
```

## Transfer the markdown file to output folder

```python
import os
import shutil

from tqdm import tqdm

src_dir= './mineru'
dst_dir= './output'

# 读取 mineru 文件下的所有文件夹
mineru_folders = os.listdir(os.path.join(os.getcwd(), src_dir))

# create a container to store all the data
mineru_files = []

# 遍历所有文件夹
for folder in tqdm(mineru_folders, desc='Scanning Mineru files'):
    # 读取文件夹下的 md 文件
    folder_files = os.listdir(os.path.join(os.getcwd(), src_dir, folder))
    for file in folder_files:
        if file.endswith('.md'):
            mineru_files.append(
                os.path.join(os.getcwd(), src_dir, folder, file)
            )

for file in tqdm(files, desc='Copying Mineru files'):
    # 将文件 copy 到 md 文件夹下
    shutil.copy(file, os.path.join(os.getcwd(), dst_dir))
```
